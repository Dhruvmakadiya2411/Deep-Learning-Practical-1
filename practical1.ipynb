## Part B – Tensor Creation

### 4. Create a Scalar Tensor

````python
import tensorflow as tf

# ==========================================================
# Task: Create a Scalar Tensor (0-D Tensor)
# ==========================================================
# tf.constant() creates an immutable (unchangeable) tensor.
# A scalar contains only a single value and has no dimensions.
scalar = tf.constant(10)

print("Scalar Tensor:")
print(scalar)
````

### 5. Create a Vector Tensor

````python
# ==========================================================
# Task: Create a Vector Tensor (1-D Tensor)
# ==========================================================
# A vector is a one-dimensional collection of values.
# Here, the vector contains four elements.

vector = tf.constant([10, 20, 30, 40])

print("Vector Tensor:")
print(vector)
````

### 6. Create a Matrix Tensor

````python
# ==========================================================
# Task: Create a Matrix Tensor (2-D Tensor)
# ==========================================================
# A matrix consists of rows and columns.
# This matrix has:
# 2 rows
# 3 columns

matrix = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matrix Tensor:")
print(matrix)
````

### 7. Create a Three-Dimensional Tensor

````python
# ==========================================================
# Task: Create a Three-Dimensional Tensor (3-D Tensor)
# ==========================================================
# A 3-D tensor can be thought of as multiple matrices stacked together.
#
# Structure:
# Layer 1
# [[1,2],
#  [3,4]]
# Layer 2
# [[5,6],
#  [7,8]]
# Total shape = (2,2,2)
# First 2  -> Number of matrices (layers)
# Second 2 -> Rows in each matrix
# Third 2  -> Columns in each matrix

tensor_3d = tf.constant([
    [
        [1, 2],
        [3, 4] ],
    [
        [5, 6],
        [7, 8] ]
])

print("Three-Dimensional Tensor:")
print(tensor_3d)
````

## Part C – Tensor Operations

### 8. Perform Tensor Addition

````python
# ==========================================================
# Task – Tensor Addition
# ==========================================================

# Create two tensors of the same shape
tensor1 = tf.constant([[1, 2],[3, 4]])

tensor2 = tf.constant([[5, 6],[7, 8]])

# tf.add() performs element-wise addition.
# Each element in tensor1 is added to the corresponding
# element in tensor2.

addition = tf.add(tensor1, tensor2)

# Display the result
print("Tensor Addition:")
print(addition)
````

### 9. Perform Tensor Subtraction

````python
# ==========================================================
# Task – Tensor Subtraction
# ==========================================================

# Create two tensors of the same shape
tensor1 = tf.constant([[1, 2], [3, 4]])

tensor2 = tf.constant([[5, 6],[7, 8]])

# tf.subtract() performs element-wise subtraction.
# Each element in tensor2 is subtracted from the
# corresponding element in tensor1.

subtraction = tf.subtract(tensor1, tensor2)

# Display the result
print("Tensor Subtraction:")
print(subtraction)
````

### 10. Perform Tensor Multiplication (Element-wise)

Note: Element-wise multiplication (`tf.multiply`) is different from matrix multiplication (`tf.matmul`), which is shown in the next task.

````python
# ==========================================================
# Task – Tensor Multiplication (Element-wise)
# ==========================================================

# Create two tensors of the same shape
tensor1 = tf.constant([[1, 2],
                       [3, 4]])

tensor2 = tf.constant([[5, 6],
                       [7, 8]])

# tf.multiply() performs element-wise multiplication.
# Each element in tensor1 is multiplied by the
# corresponding element in tensor2.

multiplication = tf.multiply(tensor1, tensor2)

# Display the result
print("Tensor Multiplication:")
print(multiplication)
````

### 11. Perform Matrix Multiplication

This uses proper matrix multiplication rules (Row x Column), not element-wise multiplication.

````python
# ==========================================================
# Task – Matrix Multiplication
# ==========================================================

# Create two matrices (2 x 2)
matrix1 = tf.constant([[1, 2],[3, 4]])

matrix2 = tf.constant([[5, 6],[7, 8]])

# tf.matmul() performs matrix multiplication.
# Unlike tf.multiply(), it follows the rules of
# matrix multiplication (Row x Column).

matrix_multiplication = tf.matmul(matrix1, matrix2)

# Display the result
print("Matrix Multiplication:")
print(matrix_multiplication)
````

### 12. Reshape an Existing Tensor

````python
# ==========================================================
# Task – Reshape an Existing Tensor
# ==========================================================

# Create a tensor of shape (2 x 2)
tensor = tf.constant([[1, 2],[3, 4]])

# Display the original tensor
print("Original Tensor:")
print(tensor)

# tf.reshape() changes the shape of a tensor
# without changing its data.
# Here, the tensor is reshaped from (2, 2) to (4, 1).

reshaped_tensor = tf.reshape(tensor, (4, 1))

# Display the reshaped tensor
print("\nReshaped Tensor:")
print(reshaped_tensor)
````

### 13. Convert a NumPy Array into a TensorFlow Tensor

````python
# ==========================================================
# Task – Convert a NumPy Array into a TensorFlow Tensor
# ==========================================================

# Import the required library
import numpy as np

# Create a NumPy array
numpy_array = np.array([[10, 20],[30, 40]])

# Display the NumPy array
print("NumPy Array:")
print(numpy_array)

# tf.convert_to_tensor() converts the NumPy array
# into a TensorFlow tensor.

tensorflow_tensor = tf.convert_to_tensor(numpy_array)

# Display the TensorFlow tensor
print("\nTensorFlow Tensor:")
print(tensorflow_tensor)
````
